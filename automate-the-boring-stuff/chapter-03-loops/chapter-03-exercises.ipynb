#### Tree Printer
Use a for loop to print a triangular pine tree of a size the user asks for. The tree branches should be printed as a number of rows of ^ characters, while the trunk should always be two # characters. For example, if the user enters 5 for the size, the program should print this:
```
Enter the tree size: 5
    ^
   ^^^
  ^^^^^
 ^^^^^^^
^^^^^^^^^
    #
    #

````python
size = int(input("Enter the size of the tree: "))
print("Enter the tree size:", size)
for row_num in range(1, size + 1):
    print(' ' * (size - row_num) + '^'* (2 * row_num - 1))
print(' ' * (size - 1) + '#')
print(' ' * (size - 1) + '#') 
````

#### Christmas Tree Printer
Instead of creating a plain tree like the one in the previous project, write a program that prints a Christmas tree with o ball ornaments randomly replacing ^ branch characters. For example, a Christmas tree of size 6 could look like this:
```
Enter the tree size: 6
     ^
    ^^o
   o^^^o
  ^o^^^o^
 ^^^^^^^^^
o^^^^^^o^oo
     #
     #

````python
import random

size = int(input("Enter the size of the christmas tree: "))
print("Enter the christmas tree size:", size)
for row_num in range(1, size + 1):
    for col_num in range(1, 2 * size):
        if col_num >= size - row_num + 1 and col_num <= size + row_num - 1:
            if random.randint(1, 4) == 1:  # Randomly decide whether to place a star (1 in 4 chance)
                print('o', end='')
            else:
                print('^', end='')
        else:
            print(' ', end='')
    print()  # Move to the next line after each row
print(' ' * (size - 1) + '#')
print(' ' * (size - 1) + '#')
````

##### Using While loop for the above program

````python
size = int(input("Enter the size of the christmas tree: "))
print("Enter the christmas tree size:", size)

row_num = 1
while row_num <= size:  # Outer while loop (replaces for row_num in range(1, size + 1))
    col_num = 1
    while col_num < 2 * size:  # Inner while loop (replaces for col_num in range(1, 2 * size))
        if col_num >= size - row_num + 1 and col_num <= size + row_num - 1:
            if random.randint(1, 4) == 1:  # Randomly decide whether to place a star (1 in 4 chance)
                print('o', end='')
            else:
                print('^', end='')
        else:
            print(' ', end='')
        col_num += 1  # Increment column counter
    print()  # Move to the next line after each row
    row_num += 1  # Increment row counter

print(' ' * (size - 1) + '#')
print(' ' * (size - 1) + '#')
````
