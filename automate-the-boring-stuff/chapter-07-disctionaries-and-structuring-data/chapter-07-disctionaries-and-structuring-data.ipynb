### Print the current chessboard

```python
def print_chessboard(board):
    squares = []
    is_white_square = True
    for y in '87654321':
        for x in 'abcdefgh':
            # print(x, y, is_white_square)  
            if x + y in board.keys(): 
                squares.append(board[x + y]) # DEBUG: Show piece on the square
            else: #
                if is_white_square: 
                    squares.append(white_square) # Adding white square
                else: 
                    squares.append(black_square) # Adding black square
            is_white_square = not is_white_square 
        is_white_square = not is_white_square

    print(board_template.format(*squares))
```

### Chess Dictionary Validator

```python
def is_valid_chess_board(board):
    valid_pieces = ['bR', 'bN', 'bB', 'bQ', 'bK', 'bP',
                    'wR', 'wN', 'wB', 'wQ', 'wK', 'wP']
    piece_count = {'bR': 2, 'bN': 2, 'bB': 2, 'bQ': 1, 'bK': 1, 'bP': 8,
                   'wR': 2, 'wN': 2, 'wB': 2, 'wQ': 1, 'wK': 1, 'wP': 8}
    for square in board.keys():
        if square[0] not in 'abcdefgh' or square[1] not in '12345678':
            return False
        piece = board[square]
        if piece not in valid_pieces:
            return False
        piece_count[piece] -= 1
    for count in piece_count.values():
        if count < 0:
            return False
    # Verify exactly one king per side
    if piece_count['bK'] != 0 or piece_count['wK'] != 0:
        return False
    return True
```

Testing the Validity of the Chess Board

```python
# Test cases for is_valid_chess_board function

# Test 1: Valid board (standard starting position)
print("Test 1 - Valid starting position:")
print(is_valid_chess_board(starting_pieces))  # Should be True

# Test 2: Valid minimal board (just kings)
print("\nTest 2 - Valid minimal board (just kings):")
minimal_board = {'e1': 'wK', 'e8': 'bK'}
print(is_valid_chess_board(minimal_board))  # Should be True


# Test 3: Invalid - invalid square notation
print("\nTest 3 - Invalid square (9z):")
invalid_square = {'e1': 'wK', 'e8': 'bK', '9z': 'wQ'}
print(is_valid_chess_board(invalid_square))  # Should be False

# Test 4: Invalid - invalid piece name
print("\nTest 4 - Invalid piece name:")
invalid_piece = {'e1': 'wK', 'e8': 'bK', 'a1': 'wX'}
print(is_valid_chess_board(invalid_piece))  # Should be False

# Test 5: Invalid - too many white pawns (9 instead of 8)
print("\nTest 5 - Too many white pawns:")
too_many_pawns = {'e1': 'wK', 'e8': 'bK'}
for i, file in enumerate('abcdefgh'):
    too_many_pawns[file + '2'] = 'wP'
too_many_pawns['a3'] = 'wP'  # Extra pawn
print(is_valid_chess_board(too_many_pawns))  # Should be False

# Test 6: Invalid - too many white rooks (3 instead of 2)
print("\nTest 6 - Too many white rooks:")
too_many_rooks = {'e1': 'wK', 'e8': 'bK', 'a1': 'wR', 'h1': 'wR', 'a2': 'wR'}
print(is_valid_chess_board(too_many_rooks))  # Should be False
```

### Fantasy Game Inventory
Say you’re creating a medieval fantasy video game. The data structure to model the player’s inventory is a dictionary whose keys are strings describing the item in the inventory and whose values are integers detailing how many of that item the player has. For example, the dictionary value {'rope': 1, 'torch': 6, 'gold coin': 42, 'dagger': 1, 'arrow': 12} means the player has one rope, six torches, 42 gold coins, and so on.

```python
def display_inventory(inventory):
    print("Inventory:")
    item_total = 0
    for item, count in inventory.items():
        print(f"{count} {item}")
        item_total += count
    print(f"Total number of items: {item_total}")

# Test case for display_inventory function

stuff = {'rope': 1, 'torch': 6, 'gold coin': 42, 'dagger': 1, 'arrow': 12, 'bread': 3, 'water': 5}
display_inventory(stuff)
```

### List-to-Dictionary Loot Conversion
Write a function named add_to_inventory(inventory, added_items). The inventory parameter is a dictionary representing the player’s inventory (as in the previous project) and the added_items parameter is a list, like dragon_loot. The add_to_inventory() function should return a dictionary that represents the player’s updated inventory. Note that the added_items list can contain multiples of the same item.

```python
def add_to_inventory(inventory, added_items):
    for item in added_items:
        inventory.setdefault(item, 0)
        inventory[item] += 1

# Test case for add_to_inventory function

inventory = {'rope': 1, 'torch': 6, 'gold coin': 42, 'dagger': 1, 'arrow': 12, 'bread': 3, 'water': 5}
dragon_loot = ['gold coin', 'dagger', 'gold coin', 'gold coin', 'ruby', 'ruby']

add_to_inventory(inventory, dragon_loot)
display_inventory(inventory)
```

### Random Weather Data Generator

Write a function named get_random_weather_data() that returns a dictionary of random weather data. 
The program should then call this function from a loop 100 times, storing the returned dictionaries in a list. Finally, it should print the list.

```python
# Random Weather Data Generator

import random
def generate_random_weather_data(num_entries):
    weather_data = {}
    for i in range(num_entries):
        temp = round(random.uniform(-50, 50), 1)  # Random temperature between -50 and 50 °C
        feels_like = round(temp + random.uniform(-10, 10), 1)  # Feels like temperature with 10 degree variation
        humidity = round(random.randint(0, 100))  # Random humidity percentage
        pressure = round(random.randint(990, 1010))  # Random pressure in hPa
        weather_data[f"Entry {i+1}"] = {
            'temperature': temp,
            'feels_like': feels_like,
            'humidity': humidity,
            'pressure': pressure
        }
    return weather_data
units = {'time': '', 'temperature': '°C', 'feels_like': '°C', 'humidity': '%', 'pressure': 'hPa'}
random_weather_data = generate_random_weather_data(100)
print("Random Weather Data:")
for entry, data in random_weather_data.items():
    formatted_data = {key: f"{value}{units[key]}" for key, value in data.items()}
    print(f"{entry}: {formatted_data}")
```

### Average-Temperature Analyzer
Add a function named get_average_temperature(weather_data) to the program in the previous practice project. This function should accept a list of the weather data dictionaries described in the previous project and return the average temperature in their 'temp' keys. To calculate the average, add all of the temperature numbers in the dictionaries and divide the result by the number of dictionaries.

The list passed to get_average_temperature() can contain any number of dictionaries but should always contain at least one. Generate a list of 100 weather dictionaries by calling get_random_weather_data(), then pass this list to get_average_temperature() and print the average it returns.

```python
# Average Temperature Calculator

def get_average_temperature(weather_data):
    total_temp = 0
    count = 0
    for data in weather_data.values():
        total_temp += data['temperature']
        count += 1
    average_temp = total_temp / count if count > 0 else 0
    return round(average_temp, 1)

average_temperature = get_average_temperature(random_weather_data)
print(f"\nAverage Temperature from Random Weather Data: {average_temperature}°C")
```

## Chess Rook Capture Predictor

```python
# Chess Rook Capture Predictor
# This function checks if a rook can capture a piece on the chessboard based on their positions.

def white_rook_can_capture(rook, board):
    rook_file = rook[0]
    rook_rank = rook[1]
    for square, piece in board.items():
        if piece.startswith('b'):  # Check only black pieces
            piece_file = square[0]
            piece_rank = square[1]
            # Check if the piece is in the same file or rank as the rook
            if piece_file == rook_file or piece_rank == rook_rank:
                return True
    return False

# Test cases for white_rook_can_capture function
test_board_1 = {'d7': 'bQ', 'd2': 'wB', 'f1': 'bP', 'a3': 'bN'}
rook_position_1 = 'd3'
print(f"Test 1 - Rook at {rook_position_1} can capture: {white_rook_can_capture(rook_position_1, test_board_1)}") 
```
