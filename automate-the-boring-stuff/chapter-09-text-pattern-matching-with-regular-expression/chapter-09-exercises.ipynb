### Extract Contact Information from Large Documents

````python
import pyperclip, re

# Create phone number regex.
phone_re = re.compile(r'''(
    # Pattern 1: International format (+358...)
    (
        \+358          # Country code
        [-\s]?         # Optional hyphen or space
        [1-9]          # First digit of prefix (not zero)
        \d{0,2}        # 0-2 more digits for prefix
        [-\s]?         # Optional hyphen or space
        \d{3,4}        # First part of number
        [-\s]?         # Optional hyphen or space
        \d{3,5}        # Second part of number
    )
    |
    # Pattern 2: Mobile numbers starting with 0
    (
        0              # Leading zero
        (
            4[05]?\d?  # 04, 040, 045, 0457
            |          # OR
            50?        # 05, 050
        )
        [-\s]?         # Optional hyphen or space
        \d{3,5}        # 3-5 digits
        [-\s]?         # Optional hyphen or space
        \d{2,5}        # 2-5 digits
    )
    |
    # Pattern 3: Other numbers starting with 0
    (
        0              # Leading zero
        [1-9]\d{0,2}   # Area code (1-3 digits)
        [-\s]?         # Optional hyphen or space
        \d{3,6}        # 3-6 digits for local number
        (
            [-\s]?      # Optional separator
            \d{1,4}     # Optional extension
        )?
    )
)''', re.VERBOSE | re.MULTILINE)

# Create email regex.
email_re = re.compile(r'''(
    [a-zA-Z0-9._%+-]+  # Username
    @  # @ symbol
    [a-zA-Z0-9.-]+  # Domain name
    (\.[a-zA-Z]{2,4})  # Dot-something
    )''', re.VERBOSE)

# Find matches in clipboard text.
text = str(pyperclip.paste())
matches = []
for group in phone_re.findall(text):
    phone_num = ' '.join([group[1], group[2], group[3]])
    matches.append(phone_num)
for group in email_re.findall(text):
    matches.append(group[0])

# Copy results to the clipboard.
if len(matches) > 0:
    pyperclip.copy('\n'.join(matches))
    print('Copied to clipboard:')
    print('\n'.join(matches))
else:
    print('No phone numbers or email addresses found.')
````

### Strong Password Detection
Write a function that uses regular expressions to make sure the password string it is passed is strong. A strong password has several rules: it must be at least eight characters long, contain both uppercase and lowercase characters, and have at least one digit. Hint: It’s easier to test the string against multiple regex patterns than to try to come up with a single regex that can validate all the rules.

````python
# Function for strong password detection

def is_strong_password(password):
    if len(password) < 8:
        return False
    if not re.search(r'[A-Z]', password):
        return False
    if not re.search(r'[a-z]', password):
        return False
    if not re.search(r'[0-9]', password):
        return False
    if not re.search(r'[@$!%*?&]', password):
        return False
    return True

# Example usage
test_passwords = ["StrongP@ssw0rd", "123456789" ,"P@55WORD", "Password123!", "Qwerty@123", "Admin@2024", "Welcome1!", "letmein"]
for pwd in test_passwords:
    if is_strong_password(pwd):
        print(f"{pwd} is a strong password.")
    else:
        print(f"{pwd} is not a strong password.")
````

### Regex Version of the strip() Method
Write a function that takes a string and does the same thing as the strip() string method. If no other arguments are passed other than the string to strip, then the function should remove whitespace characters from the beginning and end of the string. Otherwise, the function should remove the characters specified in the second argument to the function.

````python
# Regex Version of the strip() Method


def regex_strip(text, chars=None):
    if chars is None:
        # Remove leading and trailing whitespace
        return re.sub(r'^\s+|\s+$', '', text)
    else:
        # Create a regex pattern to match the specified characters
        pattern = f'^[{re.escape(chars)}]+|[{re.escape(chars)}]+$' 
        return re.sub(pattern, '', text)
    
# Example usage
print(regex_strip("   Hello, World!   "))
print(regex_strip("$$$Hello, World!$$$", "$"))
````

### Hashtag-Finding Regex
Create a regex that can find social media hashtags. For the purposes of this project, a “hashtag” pattern begins with a # character followed by one or more alphanumeric characters (letters, numbers, or underscores). Write a function named get_hashtags(sentence) that takes a string argument and returns a list of the hashtags. ````python
# Hastag-Finder

def find_hashtags(text):
    hashtag_re = re.compile(r'#[^\s!,.?;:()]+')
    return hashtag_re.findall(text)

# Prompt the user for input and find hashtags
user_input = input("Enter a sentence: ")
hashtags = find_hashtags(user_input)
if hashtags:
    print("Hashtags found:", hashtags)
else:    print("No hashtags found.")
````

### Version 1: Simple Capital-Letter Splitting
```python
formatted_text = re.sub(r'(\d)([A-Z]\.?\s?)', r'\1\n\2', mashed_text)
processed = re.sub(r'^(.*)(\d+)$', r'\1,\2', line.strip())
```
**What This Code Does**  
The first regex looks for a digit followed by a capital letter (optionally with a dot or space) and inserts a line break between them. The second regex processes each line by capturing everything before the final digits and replacing the last space with a comma.

**The Problem**  
This approach assumed every new speaker's name starts with a capital letter. It failed on:
- Non-Latin scripts (Japanese, Cyrillic, etc.) like 土屋 陽介
- Names beginning with symbols such as @torufurukawa
- Digits inside usernames (e.g., tai271828) caused incorrect splits

**Resulting Errors**  
- Created false line breaks inside legitimate entries
- Produced corrupted entries like A. Jesse Jiryu Davis1,3 (where 13 was split into 1 and 3)
- Generated 12,386 speakers – missing many valid entries

````python
import re, pyperclip


# Step 1: Get the mashed data from clipboard
mashed_text = pyperclip.paste()

# Step 2: Add line breaks between entries
# This regex finds where a digit is followed by a capital letter
# We need to be careful with names that might have numbers or capital letters in them
formatted_text = re.sub(r'(\d)([A-Z]\.?\s?)', r'\1\n\2', mashed_text)

# Step 3: Split into lines and store in 'speakers' variable
speakers = formatted_text.splitlines()

# Step 4: Process each line using (.*) to capture names with non-English characters
processed_lines = []
for line in speakers:
    if line.strip():  # Skip empty lines
        # Using (.*) to capture everything up to the digits at the end
        # This follows the project's instruction exactly
        processed = re.sub(r'^(.*)(\d+)$', r'\1,\2', line.strip())
        processed_lines.append(processed)

# Step 5: Save to CSV
with open('speakers.csv', 'w', encoding='utf-8') as file:
    file.write('\n'.join(processed_lines))

print(f"CSV file created with {len(processed_lines)} speakers")
print("Preview of first 5 lines:")
for line in processed_lines[:5]:
    print(line)
````

### Version 2: Direct Extraction with re.findall
```python
entries = re.findall(r'([^\d]+)(\d+)', mashed_text)
```
**What This Code Does**  
Instead of inserting line breaks, this version directly extracts speaker entries by finding patterns of one or more non-digit characters followed by one or more digits. This handles Unicode and symbols better than Version 1.

**The Improvement**  
- Successfully captures names starting with any character (Latin, Cyrillic, Japanese, @, etc.)
- No longer assumes capital letters for name boundaries
- Produced cleaner output with 12,792 speakers

**The New Problem**  
The pattern fails when digits appear inside the name itself:
- Taihsiang Ho (tai271828),2 splits into:  
  Entry 1: Taihsiang Ho (tai + 271828  
  Entry 2: ), + 2
- Takahiro Kubo (icoxfog417),1 splits into:  
  Entry 1: Takahiro Kubo (icoxfog + 417  
  Entry 2: ), + 1

This creates phantom entries and corrupts names with parenthetical usernames, artificially inflating the speaker count.

````python
import re
import pyperclip

# Get the mashed data from clipboard
mashed_text = pyperclip.paste()

# Find all speaker entries: name (non-digits) followed by digits
entries = re.findall(r'([^\d]+)(\d+)', mashed_text)

# Clean up names (strip extra commas/spaces) and build CSV lines
csv_lines = []
for name, count in entries:
    # Remove any trailing punctuation/whitespace that might have been captured
    clean_name = name.rstrip(' ,')
    csv_lines.append(f"{clean_name},{count}")

# Save to CSV
with open('speakers.csv', 'w', encoding='utf-8') as f:
    f.write('\n'.join(csv_lines))

print(f"Extracted {len(csv_lines)} speakers. Preview:")
for line in csv_lines[:10]:
    print(line)
````

### Version 3: Lookahead-Based Intelligent Splitting
```python
split_pattern = re.compile(r'(\d+)(?=[^\W\d_]|@|$)', re.UNICODE)
```
**What This Code Does**  
This version uses a lookahead assertion to insert line breaks only after a digit sequence that is genuinely the end of an entry. The pattern `(?=[^\W\d_]|@|$)` checks what comes after the digits:
- `[^\W\d_]` – matches any letter (in any script, including Unicode)
- `@` – matches the @ symbol (for Twitter handles)
- `$` – matches the end of the string

**Why This Works**  
Digits that belong to a username are never followed by a letter or @ – they are followed by `)`, `,`, or other punctuation. By only splitting when the next character could start a new entry, false splits are avoided.

**Edge Cases Handled**  
- Toru Furukawa1@torufurukawa,1 → splits correctly after each count
- Tsz Kiu Pang1土屋 陽介,1 → splits after 1 (followed by Japanese 土)
- Taihsiang Ho (tai271828),2 → no split at 271828 (followed by `)`), splits only at final 2
- Takahiro Kubo (icoxfog417),1 → no split at 417 (followed by `)`), splits only at final 1

**Result**  
- Accurate line breaks only at genuine entry boundaries
- Names with parenthetical usernames remain intact
- Correct speaker count (lower than Version 2's inflated count)
- Properly formatted CSV with clean name-count pairs

````python
import re
import pyperclip

mashed_text = pyperclip.paste()

# Step 1: Insert a newline after every talk count
# Pattern: one or more digits, followed by a letter (Unicode), '@', or end-of-string
split_pattern = re.compile(r'(\d+)(?=[^\W\d_]|@|$)', re.UNICODE)

parts = []
last_end = 0
for m in split_pattern.finditer(mashed_text):
    start, end = m.span()
    parts.append(mashed_text[last_end:end])   # up to and including the digits
    parts.append('\n')
    last_end = end
parts.append(mashed_text[last_end:])          # remainder after last split
formatted_text = ''.join(parts)

# Step 2: Split into lines
speakers = formatted_text.splitlines()

# Step 3: Process each line to CSV format (name,count)
processed_lines = []
for line in speakers:
    line = line.strip()
    if not line:
        continue
    # Greedy (.*) captures everything before the final digits
    m = re.match(r'^(.*)(\d+)$', line)
    if m:
        name = m.group(1).rstrip(', ')   # remove any trailing comma/space
        count = m.group(2)
        processed_lines.append(f"{name},{count}")
    else:
        # Fallback (shouldn't happen)
        processed_lines.append(line)

# Step 4: Save to CSV
with open('speakers.csv', 'w', encoding='utf-8') as f:
    f.write('\n'.join(processed_lines))

print(f"Extracted {len(processed_lines)} speakers. Preview:")
for line in processed_lines[:10]:
    print(line)
````

## Laugh Score
We can scientifically measure how funny a joke is based on the length of the text-based laughing response. For example, a joke that elicits the response “Hahaha” is objectively funnier than a joke that gets only a “Haha” response. A joke that provokes a “HAHAaaHAhhAHAHA” response is a very funny joke. Let’s write a function called laugh_score(laugh) that uses a regular expression to identify and measure the length of laughing specified by the laugh string argument. A text-based laugh is defined as beginning with ha, then consisting of any number of consecutive h or a characters. Both lowercase and uppercase characters are acceptable. If there are multiple laughs in a string, count only the first one.

To write the function, you can complete the following template:
```

````python
# Laugh Score Calculator

def laugh_score(laugh):
    laugh_re = re.compile(r'(?i)ha[h|a]*') # Case-insensitive match for 'ha' followed by any number of 'h' or 'a'
    match = laugh_re.search(laugh)
    if match:
        return len(match.group())
    else:
        return 0
    
# Example usage
assert laugh_score('abcdefg') == 0
assert laugh_score('h') == 0
assert laugh_score('ha') == 2
assert laugh_score('HA') == 2
assert laugh_score('hahaha') == 6
assert laugh_score('ha ha ha') == 2
assert laugh_score('haaaaa') == 6
assert laugh_score('ahaha') == 4
assert laugh_score('Harry said Hahaha') == 2
print("All tests passed!")
````

### Word Twister—ordW wisterT
Write a program that “twists” the words in a string. For example, calling twist_words('Hello world! How are you? I am fine.') returns 'oHell dworl! wHo ear uyo? I ma efin.' To do so, the sub() method for Pattern objects can move the last letter of every word in a string to the front of the word.

As arguments, the sub() method accepts a regex of the pattern to match, a string to replace the matches with, and the string to search for matches. Your regex should use the \b shorthand character class for word boundaries. For example, the regex ```\b[AEIOUaeiou]\w*\b``` would match every word that begins with an uppercase or lowercase vowel.

The regex should also use parentheses to put the first letter of each word in one group and the remaining letters in a second group. This way, the second argument can include the \1 and \2 back references to reorder these two groups.

````python
# Write a program that “twists” the words in a string

import re

def twist_words(s):
    # move the last letter of every word to the front
    return re.sub(r'\b(\w)(\w*)(\w)\b', r'\3\1\2', s) # \b: word boundary, (\w): first letter, (\w*): middle letters, (\w): last letter

print(twist_words('Hello world! How are you? I am fine.'))  # oHell dworl! wHo ear uyo? I ma efin.
print(twist_words('Python is great!'))  
````
