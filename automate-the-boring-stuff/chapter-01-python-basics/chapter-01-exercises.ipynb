#### Which of the two operators in the following expressions is evaluated first according to Python’s order of operation rules?

12. (4 + 5) * 6

13. 2 ** 3 + 1

14. 1 + 2 ** 3

15. (1 + 2) ** 3

16. 2 + 4 + 6

```python
# Question 12: (4 + 5) * 6
print("12. (4 + 5) * 6 =", (4 + 5) * 6)
print("    Without parentheses: 4 + 5 * 6 =", 4 + 5 * 6)
print("    This proves parentheses ( ) are evaluated first\n")

# Question 13: 2 ** 3 + 1
print("13. 2 ** 3 + 1 =", 2 ** 3 + 1)
print("    If addition was first: 2 ** (3 + 1) =", 2 ** (3 + 1))
print("    This proves exponentiation ** is before addition +\n")

# Question 14: 1 + 2 ** 3  
print("14. 1 + 2 ** 3 =", 1 + 2 ** 3)
print("    Same as Q13 - exponentiation first\n")

# Question 15: (1 + 2) ** 3
print("15. (1 + 2) ** 3 =", (1 + 2) ** 3)
print("    Without parentheses: 1 + 2 ** 3 =", 1 + 2 ** 3)
print("    Parentheses force addition before exponent\n")

# Question 16: 2 + 4 + 6
print("16. 2 + 4 + 6 =", 2 + 4 + 6)
print("    Addition is left-to-right: (2 + 4) + 6 =", (2 + 4) + 6)
print("    2 + (4 + 6) =", 2 + (4 + 6))
print("    Same result - addition is associative")
```

### Python Order of Operations (PEMDAS)

From the tests above, I can confirm Python follows:

1. **P**arentheses: `( )` - highest precedence
2. **E**xponentiation: `**` 
3. **M**ultiplication: `*` and **D**ivision: `/`, `//`, `%`
4. **A**ddition: `+` and **S**ubtraction: `-` - lowest precedence

### Rectangle Printer

Write a program that prints a rectangle of capital O characters. For example, the following rectangle has a width of eight and a height of five:
