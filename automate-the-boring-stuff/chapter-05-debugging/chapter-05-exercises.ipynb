### Raising Exceptions
Write the code that raises an Exception error with the error message 'An error happened. This error message is vague and unhelpful.'

````python
print("Enter your height:")
height = input()
if type(height) == int:
    print("You entered an integer.")
else:
    raise Exception("An error happened. This error message is vague and unhelpful.")
````

````python
def get_name():
    print('Enter your name:')
    name = input()
    if name == '':
        raise Exception('You did not enter a name.')

    return name

try:
    name = get_name()
except:
    name = 'Guido'

print('Hello,', name)
````

In this code, the raise Exception is caught by the **bare except**: block, which then assigns name = 'Guido' instead of letting the program crash.
