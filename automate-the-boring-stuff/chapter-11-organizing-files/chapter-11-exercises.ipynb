#### Selectively Copying
Write a program that walks through a folder tree and searches for files with a certain file extension (such as .pdf or .jpg). Copy these files from their current location to a new folder.

````python
# Program that walks through a folder tree searches for files with a certain file extension and copies those files to a new location.

import os, shutil, pathlib

# Get the folder to search from the user. And make sure it exists and that if the user
# enters a path with a trailing slash, that it is removed or it there is inverted.
while True:
    search_folder = input('Enter the folder to search: ')
    search_folder = os.path.normpath(search_folder)
    if os.path.exists(search_folder):
        break
    print('The folder does not exist. Please try again.')

# Get the file extension to search for from the user. And make sure it is valid.
while True:
    file_extension = input('Enter the file extension to search for (e.g. .txt): ')
    if file_extension.startswith('.'):
        break
    print('The file extension must start with a dot. Please try again.')

# Get the destination folder from the user. And create a new one by asking the user where to create it.
while True:
    destination_folder = input('Enter the destination folder to copy the files to: ')
    destination_folder = os.path.normpath(destination_folder)
    if os.path.exists(destination_folder):
        break
    create_folder = input('The folder does not exist. Do you want to create it? (y/n): ')
    if create_folder.lower() == 'y':
        new_folder_path = input('Enter the path where you want to create the new folder: ')
        new_folder_path = os.path.normpath(new_folder_path)
        if os.path.exists(new_folder_path):
            new_folder_name = input('Enter the name of the new folder: ')
            new_folder_full_path = os.path.join(new_folder_path, new_folder_name)
            try:
                os.makedirs(new_folder_full_path)
                destination_folder = new_folder_full_path
                print(f'Folder created at {new_folder_full_path}.')
                break
            except Exception as e:
                print(f'An error occurred while creating the folder: {e}')
        else:
            print('The path for the new folder does not exist. Please try again.')
    else:
        print('Please enter a valid destination folder or choose to create a new one.')

# Walk through the folder tree and copy files with the specified extension to the destination folder.
for foldername, subfolders, filenames in os.walk(search_folder):
    for filename in filenames:
        if filename.endswith(file_extension):
            source_file = os.path.join(foldername, filename)
            destination_file = os.path.join(destination_folder, filename)
            try:
                shutil.copy2(source_file, destination_file)
                print(f'Copied: {source_file} to {destination_file}')
            except Exception as e:
                print(f'An error occurred while copying {source_file}: {e}')
print('File copying complete.')
````

#### Deleting Unneeded Files
It’s not uncommon for a few unneeded but humongous files or folders to take up the bulk of the space on your hard drive. If you’re trying to free up room on your computer, it’s more effective to identify the largest unneeded files first.

Write a program that walks through a folder tree and searches for exceptionally large files or folders—say, ones that have a file size of more than 100MB. (Remember that, to get a file’s size, you can use os.path.getsize() from the os module.) Print these files with their absolute path to the screen.

````python
# Program that walks through a folder and prints out files with their extensions and sizes > 10MB.
# The files are grouped by their extensions and sorted by size in descending order.
# Get the folder to search from the user. And make sure it exists and that if the user
# enters a path with a trailing slash, that it is removed or it there is inverted.

while True:
    search_folder = input('Enter the folder to search: ')
    search_folder = os.path.normpath(search_folder)
    if os.path.exists(search_folder):
        break
    print('The folder does not exist. Please try again.')

# Walk through the folder tree and print out files with their extensions and sizes > 10MB.
files_by_extension = {}
for foldername, subfolders, filenames in os.walk(search_folder):
    for filename in filenames:
        file_path = os.path.join(foldername, filename)
        file_size = os.path.getsize(file_path)
        if file_size > 10 * 1024 * 1024:  # 10MB in bytes
            file_extension = pathlib.Path(filename).suffix
            if file_extension not in files_by_extension:
                files_by_extension[file_extension] = []
            files_by_extension[file_extension].append((filename, file_size))

# Print the files grouped by their extensions and sorted by size in descending order.
for extension, files in files_by_extension.items():
    print(f'Extension: {extension}')
    sorted_files = sorted(files, key=lambda x: x[1], reverse=True)
    for filename, size in sorted_files:
        print(f'  {filename} - {size / (1024 * 1024):.2f} MB')
````

#### Duplicate Filename Finder
Let’s say your coworker has a massive project consisting of hundreds of .txt files in different folders on their computer, but they want to make sure the filenames are all unique so that they don’t accidentally copy over any of them. Create a function named find_dup_filenames(folder) to find files with the same filename in different subfolders of folder. This function should return a dictionary whose keys are filenames and whose values are lists of absolute paths to the files. For example, let’s say the following files exist on your computer:
```
C:\Users\Al\spam.txt
C:\Users\Al\eggs.txt
C:\Users\Al\subfolder1\spam.txt
```
Calling find_dup_filenames(r'C:\Users\Al') would return this dictionary:
```
{'spam.txt': ['C:\\Users\\Al\\spam.txt', 'C:\\Users\\Al\\subfolder1\\spam.txt']}
```
Your program should call this function, then go through the returned dictionary to print duplicate filenames. Print the filename first, then each of the absolute filepaths with four spaces of indentation, so that the output looks like this:
```
spam.txt
    C:\Users\Al\spam.txt
    C:\Users\Al\subfolder1\spam.txt
```
You don’t need to test this program on your computer’s root folder; running it on every file on your computer would take too long. Instead, select your home folder or another folder containing only a handful of files and subfolders. Once you know your program works correctly, run it on your root folder to find all duplicate filenames on your computer.

````python
# Program that finds duplicate filenames in a folder tree and prints out the full paths of the duplicate files.
# Get the folder to search from the user. And make sure it exists and that if the user
# enters a path with a trailing slash, that it is removed or it there is inverted.

while True:
    search_folder = input('Enter the folder to search: ')
    search_folder = os.path.normpath(search_folder)
    if os.path.exists(search_folder):
        break
    print('The folder does not exist. Please try again.')

# Walk through the folder tree and find duplicate filenames and ignore the folder path "E:\VS Code".
filenames_dict = {}
for foldername, subfolders, filenames in os.walk(search_folder):
    if foldername == 'E:\\VS Code':
        continue
    for filename in filenames:
        if filename not in filenames_dict:
            filenames_dict[filename] = []
        full_path = os.path.join(foldername, filename)
        filenames_dict[filename].append(full_path) 

# Print out the filenames first and then full paths of the duplicate files with 4 spaces of indentation.
for filename, paths in filenames_dict.items():
    if len(paths) > 1:
        print(f'Filename: {filename}')
        for path in paths:
            print(f'    {path}')
````

#### ZIP File Folder Extractor
Let’s say your coworker has thousands of ZIP files, but they only want to extract files in a particularly named folder inside the ZIP file. Write a function named extract_in_folder(zip_filename, folder) that takes the string of a ZIP filename and a string of a folder name. The function should extract only the files in that folder of the ZIP file to the current working directory. For example, if eggs.zip contains the files data1.txt, spam/data2.txt, spam/data3.txt, and bacon/data4.txt and you call extract_in_folder('eggs.zip', 'spam'), the function should extract data2.txt and data3.txt only.

````python
# Program that extracts files from a ZIP archive to a specified folder, but only if the files are inside a specific folder within the ZIP archive.

import zipfile

def extract_in_folder(zip_filename, folder):
    with zipfile.ZipFile(zip_filename, 'r') as zip_file:
        for file_info in zip_file.infolist():
            # Extract files inside the target folder or individual files if folder is empty
            if folder and file_info.filename.startswith(folder + '/'):
                zip_file.extract(file_info)
            elif not folder and '/' not in file_info.filename:
                zip_file.extract(file_info)

# Get the ZIP archive filename from the user. And make sure it exists and that if the user
# enters a path with a trailing slash, that it is removed or it there is inverted.

while True:
    print('Enter the ZIP archive path: ')
    zip_filepath = input()
    zip_file = os.path.normpath(zip_filepath)
    if os.path.exists(zip_file):
        break
    print('The ZIP archive does not exist. Please try again.')

# Get the folder within the ZIP archive to extract from the user. And make sure it is valid.

while True:
    folder = input('Enter the folder within the ZIP archive to extract from: ')
    if folder:
        break
    print('The folder cannot be empty. Please try again.')

# Get the destination folder from the user. And make sure it exists and that if the user
# enters a path with a trailing slash, that it is removed or it there is inverted.

while True:
    destination_folder = input('Enter the destination folder to extract the files to: ')
    destination_folder = os.path.normpath(destination_folder)
    if os.path.exists(destination_folder):
        extract_in_folder(zip_file, folder)
        print('Files extracted successfully.')
        break
    print('The destination folder does not exist. Please try again.')
````

````python
import os
import zipfile

def extract_in_folder(zip_filename, folder, dest):
    """
    Extract members of `zip_filename` into the directory `dest`.

    If `folder` is non‑empty, only archive entries whose name begins with
    “folder/” are extracted.  Trailing slashes on `folder` are ignored.
    If `folder` is the empty string, only entries stored at the root of the
    archive (i.e. names containing no slash) are extracted.  Directory
    entries themselves are skipped; extracting the files recreates the
    needed subdirectories automatically.
    """
    with zipfile.ZipFile(zip_filename, 'r') as zf:
        for info in zf.infolist():
            name = info.filename
            # skip explicit directory records
            if name.endswith('/'):
                continue
            if folder: 
                prefix = folder.rstrip('/') + '/' 
                if name.startswith(prefix): 
                    zf.extract(info, dest)
            else:
                if '/' not in name:
                    zf.extract(info, dest)


def ask_for_existing_path(prompt):
    """Prompt repeatedly until the user supplies an existing path."""
    while True:
        p = input(prompt).strip()
        p = os.path.normpath(p)
        if os.path.exists(p):
            return p
        print('The path does not exist. Please try again.')

def ask_for_folder_name():
    """
    Ask which folder inside the zip to extract.

    An empty response is allowed and means “extract only files at the top
    level of the archive”.
    """
    return input(
        'Enter the folder within the ZIP archive to extract from '
        '(leave blank for root files): '
    ).strip()

def main():
    # ask for zip file path
    zip_path = ask_for_existing_path('Enter the ZIP archive path: ')

    # ask for folder name (may be an empty string)
    folder = ask_for_folder_name()

    # ask for destination directory, create it if the user agrees
    while True:
        dest = input('Enter the destination folder to extract the files to: ')
        dest = os.path.normpath(dest)
        if os.path.exists(dest):
            break
        create = input('Destination does not exist. Create it? (y/n): ').lower()
        if create == 'y':
            try:
                os.makedirs(dest, exist_ok=True)
                break
            except OSError as e:
                print(f'Unable to create destination: {e}')
        else:
            print('Please enter or create a valid destination folder.')

    extract_in_folder(zip_path, folder, dest)
    print('Files extracted successfully.')

if __name__ == '__main__':
    main()
````
