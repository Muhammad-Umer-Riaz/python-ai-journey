#### Selectively Copying
Write a program that walks through a folder tree and searches for files with a certain file extension (such as .pdf or .jpg). Copy these files from their current location to a new folder.

```python
# Program that walks through a folder tree searches for files with a certain file extension and copies those files to a new location.

import os, shutil, pathlib

# Get the folder to search from the user. And make sure it exists and that if the user
# enters a path with a trailing slash, that it is removed or it there is inverted.
while True:
    search_folder = input('Enter the folder to search: ')
    search_folder = os.path.normpath(search_folder)
    if os.path.exists(search_folder):
        break
    print('The folder does not exist. Please try again.')

# Get the file extension to search for from the user. And make sure it is valid.
while True:
    file_extension = input('Enter the file extension to search for (e.g. .txt): ')
    if file_extension.startswith('.'):
        break
    print('The file extension must start with a dot. Please try again.')

# Get the destination folder from the user. And create it if it doesn't exist.
while True:
    destination_folder = input('Enter the destination folder: ')
    destination_folder = os.path.normpath(destination_folder)
    if not os.path.exists(destination_folder):
        os.makedirs(destination_folder)
    break
    print('The destination folder does not exist. Please try again.')
```
